# EventArena Intent Classifier

This notebook trains a text intent classifier using synthetic, hand-authored examples. This is a demonstration dataset, not real user data.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
import joblib
```

```python
data = pd.read_csv('eventarena_synthetic_intents.csv')
data.head(), data['intent'].value_counts()
```

```python
X_train, X_test, y_train, y_test = train_test_split(data['text'], data['intent'], test_size=0.22, random_state=42, stratify=data['intent'])
```

```python
model = Pipeline([('tfidf', TfidfVectorizer(ngram_range=(1,2), lowercase=True, strip_accents='unicode')), ('classifier', LogisticRegression(max_iter=2000, class_weight='balanced', random_state=42))])
model.fit(X_train, y_train)
```

```python
pred = model.predict(X_test)
print('Accuracy:', accuracy_score(y_test, pred))
print(classification_report(y_test, pred, zero_division=0))
```

```python
joblib.dump(model, 'intent_classifier.joblib')
print('Saved intent_classifier.joblib')
```

## Important limitations

Synthetic examples are useful for a reproducible prototype but may not represent real user language. Test with genuinely collected, consented utterances where permitted. Review per-class precision/recall and calibrate the fallback threshold on a validation set. The app currently uses 0.30 as a demonstration threshold, not a validated optimum.
